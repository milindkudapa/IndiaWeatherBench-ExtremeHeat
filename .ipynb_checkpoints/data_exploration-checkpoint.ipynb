# IndiaWeatherBench Data Exploration

This notebook provides a comprehensive exploration of the IndiaWeatherBench dataset, including:
- Dataset structure and metadata
- Variable statistics and distributions
- Temporal patterns and seasonality
- Spatial patterns and geographic coverage
- Data quality checks (missing values, outliers)
- Variable correlations
- Visualizations

## Dataset Overview
- **Spatial Resolution**: 256×256 grid over India
- **Temporal Resolution**: 6-hourly data
- **Geographic Coverage**: ~6-36.7°N, ~66.5-97.2°E
- **Time Period**: 2000-2019 (train: 2000-2017, val: 2018, test: 2019)
- **Total Variables**: 43 (37 atmospheric + 6 ERA5-Land)


```python
# Import necessary libraries
import os
import json
import h5py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from glob import glob
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# Set style for better visualizations
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Set paths
PROJECT_ROOT = "/burg-archive/home/mck2199/ML-Project"
DATA_ROOT = os.path.join(PROJECT_ROOT, "data/indibench_h5")
NORM_PARAMS_PATH = os.path.join(DATA_ROOT, "norm_params.json")

print(f"Data root: {DATA_ROOT}")
print(f"Normalization params: {NORM_PARAMS_PATH}")
```

## 1. Dataset Structure and Metadata


```python
# Load normalization parameters
with open(NORM_PARAMS_PATH, 'r') as f:
    norm_params = json.load(f)

# Get all variables from normalization params
all_variables = list(norm_params['mean'].keys())
print(f"Total variables: {len(all_variables)}")
print(f"\nVariables: {all_variables}")

# Load a sample file to understand structure
sample_file = os.path.join(DATA_ROOT, "train/2000-01-01_00.h5")
print(f"\nSample file: {sample_file}")

with h5py.File(sample_file, 'r') as f:
    print(f"\nHDF5 file keys: {list(f.keys())}")
    print(f"\nFile structure:")
    for key in f.keys():
        print(f"  {key}: shape={f[key].shape}, dtype={f[key].dtype}")
    
    # Get spatial dimensions from first variable
    first_var = list(f.keys())[0]
    spatial_shape = f[first_var].shape
    print(f"\nSpatial dimensions: {spatial_shape}")
```

```python
# Count files in each split
train_files = glob(os.path.join(DATA_ROOT, "train/*.h5"))
val_files = glob(os.path.join(DATA_ROOT, "val/*.h5"))
test_files = glob(os.path.join(DATA_ROOT, "test/*.h5"))

print(f"Training files: {len(train_files)}")
print(f"Validation files: {len(val_files)}")
print(f"Test files: {len(test_files)}")
print(f"Total files: {len(train_files) + len(val_files) + len(test_files)}")

# Extract dates from filenames
def extract_dates(file_list):
    dates = []
    for f in file_list:
        filename = os.path.basename(f)
        date_str = filename.split('_')[0]
        dates.append(pd.to_datetime(date_str))
    return sorted(dates)

train_dates = extract_dates(train_files)
val_dates = extract_dates(val_files)
test_dates = extract_dates(test_files)

print(f"\nDate ranges:")
print(f"  Train: {train_dates[0]} to {train_dates[-1]}")
print(f"  Val:   {val_dates[0]} to {val_dates[-1]}")
print(f"  Test:  {test_dates[0]} to {test_dates[-1]}")
```

## 2. Variable Categories and Descriptions


```python
# Categorize variables
surface_vars = ['TMP', 'UGRD', 'VGRD', 'PRMSL']
hgt_vars = [v for v in all_variables if v.startswith('HGT')]
tmp_prl_vars = [v for v in all_variables if v.startswith('TMP_prl')]
ugrd_prl_vars = [v for v in all_variables if v.startswith('UGRD_prl')]
vgrd_prl_vars = [v for v in all_variables if v.startswith('VGRD_prl')]
rh_vars = [v for v in all_variables if v.startswith('RH')]
precip_vars = [v for v in all_variables if 'APCP' in v]
era5_soil = ['swvl1', 'swvl2']
era5_flux = ['slhf', 'sshf']
era5_veg = ['lai_hv', 'lai_lv']

all_categorized = (surface_vars + hgt_vars + tmp_prl_vars + ugrd_prl_vars + 
                   vgrd_prl_vars + rh_vars + precip_vars + era5_soil + era5_flux + era5_veg)
other_vars = [v for v in all_variables if v not in all_categorized]

variable_categories = {
    'Surface Variables': surface_vars,
    'Geopotential Height': hgt_vars,
    'Temperature (Pressure Levels)': tmp_prl_vars,
    'U-Wind (Pressure Levels)': ugrd_prl_vars,
    'V-Wind (Pressure Levels)': vgrd_prl_vars,
    'Relative Humidity': rh_vars,
    'Precipitation': precip_vars,
    'ERA5-Land Soil Moisture': era5_soil,
    'ERA5-Land Heat Fluxes': era5_flux,
    'ERA5-Land Vegetation': era5_veg
}

# Add 'Other' category if there are uncategorized variables
if other_vars:
    variable_categories['Other'] = other_vars

print("Variable Categories:")
for category, vars_list in variable_categories.items():
    print(f"\n{category} ({len(vars_list)} variables):")
    for var in vars_list:
        print(f"  - {var}")

# Create a summary DataFrame
var_info = []
for var in all_variables:
    category = next((cat for cat, vars_list in variable_categories.items() if var in vars_list), 'Other')
    var_info.append({
        'Variable': var,
        'Category': category,
        'Mean': norm_params['mean'][var],
        'Std': norm_params['std'][var],
        'Min': norm_params['mean'][var] - 3 * norm_params['std'][var],  # Approximate
        'Max': norm_params['mean'][var] + 3 * norm_params['std'][var]   # Approximate
    })

var_df = pd.DataFrame(var_info)
print(f"\n\nVariable Summary DataFrame:")
print(var_df.head(10))
```

## 3. Load Sample Data and Basic Statistics


```python
# Load a sample of data files for analysis
def load_sample_data(file_paths, variables, max_files=100):
    """Load data from multiple HDF5 files"""
    data_dict = {var: [] for var in variables}
    
    for file_path in file_paths[:max_files]:
        try:
            with h5py.File(file_path, 'r') as f:
                for var in variables:
                    if var in f.keys():
                        data_dict[var].append(f[var][()])
        except Exception as e:
            print(f"Error loading {file_path}: {e}")
            continue
    
    # Stack arrays
    for var in variables:
        if data_dict[var]:
            data_dict[var] = np.stack(data_dict[var], axis=0)
        else:
            data_dict[var] = None
    
    return data_dict

# Load sample from training data
print("Loading sample data (this may take a moment)...")
sample_data = load_sample_data(train_files, all_variables[:10], max_files=50)  # Load first 10 vars, 50 files

print(f"\nLoaded data shapes:")
for var, data in sample_data.items():
    if data is not None:
        print(f"  {var}: {data.shape}")
```

```python
# Compute basic statistics for each variable
def compute_statistics(data_dict, norm_params):
    """Compute comprehensive statistics for each variable"""
    stats = []
    
    for var, data in data_dict.items():
        if data is None:
            continue
        
        # Flatten spatial dimensions for statistics
        flat_data = data.flatten()
        
        stats.append({
            'Variable': var,
            'Count': len(flat_data),
            'Mean': float(np.mean(flat_data)),
            'Std': float(np.std(flat_data)),
            'Min': float(np.min(flat_data)),
            'Max': float(np.max(flat_data)),
            'Median': float(np.median(flat_data)),
            'Q25': float(np.percentile(flat_data, 25)),
            'Q75': float(np.percentile(flat_data, 75)),
            'Norm_Mean': norm_params['mean'][var],
            'Norm_Std': norm_params['std'][var],
            'Missing': int(np.isnan(flat_data).sum()),
            'Inf': int(np.isinf(flat_data).sum())
        })
    
    return pd.DataFrame(stats)

stats_df = compute_statistics(sample_data, norm_params)
print("Basic Statistics:")
print(stats_df.to_string(index=False))
```

## 4. Geographic Coverage and Spatial Grid


```python
# Extract latitude and longitude from normalization params
lats = np.array(norm_params['lat'])
lons = np.array(norm_params['lon'])

print(f"Latitude range: {lats.min():.2f}°N to {lats.max():.2f}°N")
print(f"Longitude range: {lons.min():.2f}°E to {lons.max():.2f}°E")
print(f"Grid size: {len(lats)} × {len(lons)} = {len(lats) * len(lons)} points")
print(f"Spatial resolution: ~{np.diff(lats).mean():.3f}° lat, ~{np.diff(lons).mean():.3f}° lon")

# Create a simple map visualization
fig, ax = plt.subplots(figsize=(10, 8))
ax.scatter(lons, lats, s=1, alpha=0.3, c='blue')
ax.set_xlabel('Longitude (°E)', fontsize=12)
ax.set_ylabel('Latitude (°N)', fontsize=12)
ax.set_title('Geographic Coverage of Dataset', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)

# Add India outline (approximate)
india_lat = [6, 36.7, 36.7, 6, 6]
india_lon = [66.5, 66.5, 97.2, 97.2, 66.5]
ax.plot(india_lon, india_lat, 'r-', linewidth=2, label='India Boundary')
ax.legend()

plt.tight_layout()
plt.show()
```

## 5. Variable Distributions


```python
# Plot distributions for key variables
key_variables = ['TMP', 'UGRD', 'VGRD', 'PRMSL', 'swvl1', 'swvl2', 'slhf', 'sshf']

# Load full data for these variables
print("Loading data for distribution analysis...")
full_sample = load_sample_data(train_files, key_variables, max_files=100)

fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

for idx, var in enumerate(key_variables):
    if var not in full_sample or full_sample[var] is None:
        continue
    
    data = full_sample[var].flatten()
    # Remove outliers for better visualization (beyond 3 std)
    mean_val = norm_params['mean'][var]
    std_val = norm_params['std'][var]
    mask = np.abs(data - mean_val) < 3 * std_val
    data_filtered = data[mask]
    
    axes[idx].hist(data_filtered, bins=50, alpha=0.7, edgecolor='black')
    axes[idx].axvline(mean_val, color='r', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.2f}')
    axes[idx].set_title(f'{var} Distribution', fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('Value', fontsize=10)
    axes[idx].set_ylabel('Frequency', fontsize=10)
    axes[idx].legend()
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. Temporal Patterns and Seasonality


```python
# Analyze temporal patterns
def load_temporal_data(file_paths, variables, max_files=500):
    """Load data with timestamps"""
    data_dict = {var: [] for var in variables}
    timestamps = []
    
    for file_path in sorted(file_paths)[:max_files]:
        try:
            filename = os.path.basename(file_path)
            date_str, time_str = filename.replace('.h5', '').split('_')
            timestamp = pd.to_datetime(f"{date_str} {time_str}:00:00")
            timestamps.append(timestamp)
            
            with h5py.File(file_path, 'r') as f:
                for var in variables:
                    if var in f.keys():
                        # Take spatial mean for temporal analysis
                        data_dict[var].append(np.mean(f[var][()]))
        except Exception as e:
            continue
    
    # Convert to arrays
    for var in variables:
        if data_dict[var]:
            data_dict[var] = np.array(data_dict[var])
    
    return pd.DataFrame(data_dict, index=timestamps)

print("Loading temporal data...")
temporal_vars = ['TMP', 'PRMSL', 'swvl1', 'swvl2']
temporal_df = load_temporal_data(train_files, temporal_vars, max_files=1000)

print(f"\nTemporal data shape: {temporal_df.shape}")
print(f"Date range: {temporal_df.index.min()} to {temporal_df.index.max()}")
print(f"\nFirst few rows:")
print(temporal_df.head())
```

```python
# Plot temporal patterns
fig, axes = plt.subplots(len(temporal_vars), 1, figsize=(15, 4*len(temporal_vars)))

for idx, var in enumerate(temporal_vars):
    if var not in temporal_df.columns:
        continue
    
    axes[idx].plot(temporal_df.index, temporal_df[var], linewidth=0.5, alpha=0.7)
    axes[idx].set_title(f'{var} Temporal Pattern (Spatial Mean)', fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('Date', fontsize=10)
    axes[idx].set_ylabel('Value', fontsize=10)
    axes[idx].grid(True, alpha=0.3)
    
    # Add monthly rolling mean
    monthly_mean = temporal_df[var].resample('M').mean()
    axes[idx].plot(monthly_mean.index, monthly_mean.values, 'r-', linewidth=2, label='Monthly Mean')
    axes[idx].legend()

plt.tight_layout()
plt.show()
```

```python
# Analyze seasonality
temporal_df['Month'] = temporal_df.index.month
temporal_df['Year'] = temporal_df.index.year

# Monthly statistics
monthly_stats = temporal_df.groupby('Month')[temporal_vars].agg(['mean', 'std'])
print("Monthly Statistics:")
print(monthly_stats)

# Plot seasonal patterns
fig, axes = plt.subplots(len(temporal_vars), 1, figsize=(12, 4*len(temporal_vars)))

for idx, var in enumerate(temporal_vars):
    if var not in temporal_df.columns:
        continue
    
    monthly_means = temporal_df.groupby('Month')[var].mean()
    monthly_stds = temporal_df.groupby('Month')[var].std()
    
    months = range(1, 13)
    axes[idx].plot(months, monthly_means.values, 'o-', linewidth=2, markersize=8, label='Mean')
    axes[idx].fill_between(months, 
                           monthly_means.values - monthly_stds.values,
                           monthly_means.values + monthly_stds.values,
                           alpha=0.3, label='±1 Std')
    axes[idx].set_title(f'{var} Seasonal Pattern', fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('Month', fontsize=10)
    axes[idx].set_ylabel('Value', fontsize=10)
    axes[idx].set_xticks(months)
    axes[idx].set_xticklabels(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
    axes[idx].grid(True, alpha=0.3)
    axes[idx].legend()

plt.tight_layout()
plt.show()
```

## 7. Spatial Patterns


```python
# Load spatial data for visualization
def load_spatial_data(file_path, variables):
    """Load spatial data from a single file"""
    data_dict = {}
    with h5py.File(file_path, 'r') as f:
        for var in variables:
            if var in f.keys():
                data_dict[var] = f[var][()]
    return data_dict

# Load a sample file
sample_file = os.path.join(DATA_ROOT, "train/2000-06-15_00.h5")  # Mid-summer sample
spatial_data = load_spatial_data(sample_file, key_variables)

print(f"Loaded spatial data from: {os.path.basename(sample_file)}")
print(f"Variables: {list(spatial_data.keys())}")
```

```python
# Plot spatial patterns
try:
    import cartopy.crs as ccrs
    import cartopy.feature as cfeature
    HAS_CARTOPY = True
except ImportError:
    HAS_CARTOPY = False
    print("Note: cartopy not available, using simple contour plots")

# Create maps for key variables
plot_vars = ['TMP', 'PRMSL', 'UGRD', 'VGRD', 'swvl1', 'swvl2']
n_vars = len(plot_vars)

fig = plt.figure(figsize=(18, 12))

for idx, var in enumerate(plot_vars):
    if var not in spatial_data:
        continue
    
    if HAS_CARTOPY:
        ax = plt.subplot(2, 3, idx+1, projection=ccrs.PlateCarree())
    else:
        ax = plt.subplot(2, 3, idx+1)
    
    # Create meshgrid for coordinates
    lon_grid, lat_grid = np.meshgrid(lons, lats)
    
    # Plot data
    if HAS_CARTOPY:
        im = ax.contourf(lon_grid, lat_grid, spatial_data[var], 
                         levels=20, transform=ccrs.PlateCarree(), cmap='viridis')
        # Add coastlines and borders
        ax.coastlines(resolution='50m', linewidth=0.5)
        ax.add_feature(cfeature.BORDERS, linewidth=0.5)
    else:
        im = ax.contourf(lon_grid, lat_grid, spatial_data[var], 
                         levels=20, cmap='viridis')
    
    # Add colorbar
    plt.colorbar(im, ax=ax, orientation='horizontal', pad=0.05, label=var)
    
    ax.set_title(f'{var} - Spatial Distribution', fontsize=12, fontweight='bold')
    ax.set_xlabel('Longitude (°E)', fontsize=10)
    ax.set_ylabel('Latitude (°N)', fontsize=10)
    if not HAS_CARTOPY:
        ax.set_xlim(lons.min(), lons.max())
        ax.set_ylim(lats.min(), lats.max())
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 8. Variable Correlations


```python
# Compute correlations between variables
correlation_vars = ['TMP', 'UGRD', 'VGRD', 'PRMSL', 'HGT500', 'RH850', 'swvl1', 'swvl2', 'slhf', 'sshf']

# Load data for correlation analysis
print("Loading data for correlation analysis...")
corr_data = load_sample_data(train_files, correlation_vars, max_files=200)

# Flatten spatial dimensions
corr_dict = {}
for var in correlation_vars:
    if var in corr_data and corr_data[var] is not None:
        corr_dict[var] = corr_data[var].flatten()

# Create DataFrame
corr_df = pd.DataFrame(corr_dict)

# Compute correlation matrix
correlation_matrix = corr_df.corr()

# Plot correlation heatmap
plt.figure(figsize=(12, 10))
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))
sns.heatmap(correlation_matrix, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title('Variable Correlation Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("\nStrongest Correlations (|r| > 0.7):")
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_val = correlation_matrix.iloc[i, j]
        if abs(corr_val) > 0.7:
            print(f"  {correlation_matrix.columns[i]} <-> {correlation_matrix.columns[j]}: {corr_val:.3f}")
```

## 9. Data Quality Checks


```python
# Check for missing values, NaN, Inf across multiple files
def check_data_quality(file_paths, variables, max_files=100):
    """Check data quality across files"""
    quality_report = {var: {'nan_count': 0, 'inf_count': 0, 'files_checked': 0} 
                     for var in variables}
    
    for file_path in file_paths[:max_files]:
        try:
            with h5py.File(file_path, 'r') as f:
                for var in variables:
                    if var in f.keys():
                        data = f[var][()]
                        quality_report[var]['nan_count'] += np.isnan(data).sum()
                        quality_report[var]['inf_count'] += np.isinf(data).sum()
                        quality_report[var]['files_checked'] += 1
        except Exception as e:
            continue
    
    return quality_report

print("Checking data quality...")
quality_report = check_data_quality(train_files, all_variables[:15], max_files=200)

quality_df = pd.DataFrame(quality_report).T
quality_df['total_values'] = quality_df['files_checked'] * 256 * 256  # Assuming 256x256 grid
quality_df['nan_pct'] = (quality_df['nan_count'] / quality_df['total_values']) * 100
quality_df['inf_pct'] = (quality_df['inf_count'] / quality_df['total_values']) * 100

print("\nData Quality Report:")
print(quality_df[['files_checked', 'nan_count', 'inf_count', 'nan_pct', 'inf_pct']].to_string())

# Check for variables with issues
problem_vars = quality_df[(quality_df['nan_pct'] > 0) | (quality_df['inf_pct'] > 0)]
if len(problem_vars) > 0:
    print(f"\n⚠️ Variables with data quality issues:")
    print(problem_vars[['nan_pct', 'inf_pct']].to_string())
else:
    print("\n✓ No data quality issues detected!")

# Initialize problem_vars as empty DataFrame if not defined
if 'problem_vars' not in locals():
    problem_vars = pd.DataFrame()
```

## 10. Normalization Parameters Analysis


```python
# Analyze normalization parameters
norm_analysis = pd.DataFrame({
    'Variable': all_variables,
    'Mean': [norm_params['mean'][v] for v in all_variables],
    'Std': [norm_params['std'][v] for v in all_variables],
    'CV': [norm_params['std'][v] / abs(norm_params['mean'][v]) if norm_params['mean'][v] != 0 else np.inf 
           for v in all_variables]
})

# Sort by coefficient of variation
norm_analysis = norm_analysis.sort_values('CV', ascending=False)

print("Normalization Parameters Analysis:")
print(f"\nVariables with highest variability (CV):")
print(norm_analysis.head(10).to_string(index=False))

print(f"\nVariables with lowest variability (CV):")
print(norm_analysis.tail(10).to_string(index=False))

# Visualize normalization parameters
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Mean values
axes[0, 0].barh(range(len(norm_analysis)), norm_analysis['Mean'].values)
axes[0, 0].set_yticks(range(len(norm_analysis)))
axes[0, 0].set_yticklabels(norm_analysis['Variable'].values, fontsize=8)
axes[0, 0].set_xlabel('Mean Value', fontsize=10)
axes[0, 0].set_title('Normalization Means', fontsize=12, fontweight='bold')
axes[0, 0].grid(True, alpha=0.3)

# Std values
axes[0, 1].barh(range(len(norm_analysis)), norm_analysis['Std'].values)
axes[0, 1].set_yticks(range(len(norm_analysis)))
axes[0, 1].set_yticklabels(norm_analysis['Variable'].values, fontsize=8)
axes[0, 1].set_xlabel('Std Value', fontsize=10)
axes[0, 1].set_title('Normalization Standard Deviations', fontsize=12, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)

# Coefficient of Variation
axes[1, 0].barh(range(len(norm_analysis)), norm_analysis['CV'].values)
axes[1, 0].set_yticks(range(len(norm_analysis)))
axes[1, 0].set_yticklabels(norm_analysis['Variable'].values, fontsize=8)
axes[1, 0].set_xlabel('Coefficient of Variation', fontsize=10)
axes[1, 0].set_title('Coefficient of Variation (Std/Mean)', fontsize=12, fontweight='bold')
axes[1, 0].grid(True, alpha=0.3)

# Mean vs Std scatter
axes[1, 1].scatter(norm_analysis['Mean'], norm_analysis['Std'], alpha=0.6)
axes[1, 1].set_xlabel('Mean', fontsize=10)
axes[1, 1].set_ylabel('Std', fontsize=10)
axes[1, 1].set_title('Mean vs Standard Deviation', fontsize=12, fontweight='bold')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 11. Pressure Level Analysis


```python
# Analyze variables across pressure levels
pressure_levels = {
    '925': ['HGT925', 'TMP_prl925', 'UGRD_prl925', 'VGRD_prl925', 'RH925'],
    '850': ['HGT850', 'TMP_prl850', 'UGRD_prl850', 'VGRD_prl850', 'RH850'],
    '700': ['HGT700', 'TMP_prl700', 'UGRD_prl700', 'VGRD_prl700', 'RH700'],
    '600': ['HGT600', 'TMP_prl600', 'UGRD_prl600', 'VGRD_prl600', 'RH600'],
    '500': ['HGT500', 'TMP_prl500', 'UGRD_prl500', 'VGRD_prl500', 'RH500'],
    '250': ['HGT250', 'TMP_prl250', 'UGRD_prl250', 'VGRD_prl250', 'RH250'],
    '50': ['HGT50', 'TMP_prl50', 'UGRD_prl50', 'VGRD_prl50', 'RH50']
}

# Extract temperature values at different pressure levels
temp_levels = {}
for level, vars_list in pressure_levels.items():
    temp_var = f'TMP_prl{level}'
    if temp_var in norm_params['mean']:
        temp_levels[level] = norm_params['mean'][temp_var]

if temp_levels:
    levels_sorted = sorted(temp_levels.keys(), key=lambda x: int(x))
    temp_values = [temp_levels[l] for l in levels_sorted]
    
    plt.figure(figsize=(8, 10))
    plt.plot(temp_values, levels_sorted, 'o-', linewidth=2, markersize=8)
    plt.xlabel('Temperature (K)', fontsize=12)
    plt.ylabel('Pressure Level (hPa)', fontsize=12)
    plt.title('Temperature Profile Across Pressure Levels', fontsize=14, fontweight='bold')
    plt.grid(True, alpha=0.3)
    plt.gca().invert_yaxis()  # Invert y-axis so higher pressure is at bottom
    plt.tight_layout()
    plt.show()
    
    print("\nTemperature at Pressure Levels:")
    for level, temp in zip(levels_sorted, temp_values):
        print(f"  {level} hPa: {temp:.2f} K")
```

## 12. ERA5-Land Variables Analysis


```python
# Focus on ERA5-Land variables
era5_vars = ['swvl1', 'swvl2', 'slhf', 'sshf', 'lai_hv', 'lai_lv']

print("ERA5-Land Variables Summary:")
era5_summary = var_df[var_df['Variable'].isin(era5_vars)]
print(era5_summary[['Variable', 'Category', 'Mean', 'Std', 'Min', 'Max']].to_string(index=False))

# Load ERA5-Land data for detailed analysis
print("\nLoading ERA5-Land data...")
era5_data = load_sample_data(train_files, era5_vars, max_files=200)

# Plot ERA5-Land variable distributions
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for idx, var in enumerate(era5_vars):
    if var not in era5_data or era5_data[var] is None:
        continue
    
    data = era5_data[var].flatten()
    
    axes[idx].hist(data, bins=50, alpha=0.7, edgecolor='black')
    axes[idx].axvline(norm_params['mean'][var], color='r', linestyle='--', 
                     linewidth=2, label=f'Mean: {norm_params["mean"][var]:.2f}')
    axes[idx].set_title(f'{var} Distribution', fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('Value', fontsize=10)
    axes[idx].set_ylabel('Frequency', fontsize=10)
    axes[idx].legend()
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Analyze relationship between soil moisture variables
if 'swvl1' in era5_data and 'swvl2' in era5_data and era5_data['swvl1'] is not None:
    swvl1_flat = era5_data['swvl1'].flatten()
    swvl2_flat = era5_data['swvl2'].flatten()
    
    plt.figure(figsize=(8, 6))
    plt.scatter(swvl1_flat[::1000], swvl2_flat[::1000], alpha=0.3, s=1)  # Subsample for visualization
    plt.xlabel('swvl1 (Layer 1: 0-7 cm)', fontsize=12)
    plt.ylabel('swvl2 (Layer 2: 7-28 cm)', fontsize=12)
    plt.title('Soil Moisture Layer 1 vs Layer 2', fontsize=14, fontweight='bold')
    plt.grid(True, alpha=0.3)
    
    # Add correlation
    corr = np.corrcoef(swvl1_flat, swvl2_flat)[0, 1]
    plt.text(0.05, 0.95, f'Correlation: {corr:.3f}', transform=plt.gca().transAxes,
             fontsize=12, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
```
